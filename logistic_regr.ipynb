```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LogisticRegression

# Length and height data for Dachshunds
dach_length = [55, 57, 64, 63, 58, 49, 54, 61]
dach_height = [30, 31, 36, 30, 33, 25, 37, 34]

# Length and height data for Jindo dogs
jin_length = [56, 47, 56, 46, 49, 53, 52, 48]
jin_height = [52, 52, 50, 53, 50, 53, 49, 54]

# Create labels: Dachshund = 0, Jindo = 1
X = np.array(
    [[l, h] for l, h in zip(dach_length + jin_length,
                          dach_height + jin_height)]
)
y = np.array([0]*len(dach_length) + [1]*len(jin_length))

# X와 y 출력하기
print("--- X (Feature data :: Matrix(2D array):16x2 ---")
print(X)

print("\n--- y (label data :: 1D Array(Vector)) ---")
print(y)

# Train logistic regression model
model = LogisticRegression()
model.fit(X, y)
```

```python
# ---- Visualization ----
plt.scatter(dach_length, dach_height, c='r', label='Dachshund')
plt.scatter(jin_length, jin_height, c='b', marker='^', label='Jindo dog')
```

```python
# ---- Visualization ----
plt.scatter(dach_length, dach_height, c='r', label='Dachshund')
plt.scatter(jin_length, jin_height, c='b', marker='^', label='Jindo dog')

# --- (1) Predict model decision line ---
# create a range of x (length)
x_values = np.linspace(min(X[:,0])-2, max(X[:,0])+2, 200)

# decision boundary: model.coef_[0][0] * x + model.coef_[0][1] * y + intercept = 0
# solve for y: y = -(w1*x + b)/w2
w1 = model.coef_[0][0]
w2 = model.coef_[0][1]
b = model.intercept_[0]

y_values = -(w1 * x_values + b) / w2

plt.plot(x_values, y_values, 'g--', linewidth=2, label='Decision Boundary')

plt.xlabel('Length')
plt.ylabel('Height')
plt.title("Dog Classification with Decision Line")
plt.legend(loc='upper right')

plt.show()
```

```python

# ---- Visualization ----
plt.figure(figsize=(8, 5))

# Dachshund dogs (red dots)
plt.scatter(
    dach_length, dach_height,
    c='red', marker='o', label='Dachshund'
)

# Jindo dogs (blue triangles)
plt.scatter(
    jin_length, jin_height,
    c='blue', marker='^', label='Jindo dog'
)

# New data (green pentagon)
newdata_length = [59]
newdata_height = [35]

plt.scatter(
    newdata_length, newdata_height,
    c='green', marker='p', s=150, label='new Data'
)

# Decision boundary (green dashed line)
x_values = np.linspace(min(X[:, 0]) - 2,
                       max(X[:, 0]) + 2, 200)

w1 = model.coef_[0][0]
w2 = model.coef_[0][1]
b = model.intercept_[0]

y_values = -(w1 * x_values + b) / w2

plt.plot(
    x_values, y_values,
    'g--', linewidth=2, label='Decision Boundary'
)

# Labels and title
plt.xlabel('Length')
plt.ylabel('Height')
plt.title('Dog size by Anvarov Samandar (student ID 202390206)')
plt.legend(loc='upper right')

plt.show()

# Predict the new dog's breed
newdata = np.array([[59, 35]])
prediction = model.predict(newdata)

if prediction[0] == 0:
    print("Prediction: Dachshund")
else:
    print("Prediction: Jindo dog")

```
